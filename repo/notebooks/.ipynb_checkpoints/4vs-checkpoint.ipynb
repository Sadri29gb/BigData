# Caso de Estudio: CityBike Lima y Big Data

**CityBike Lima** esta empresa brinda un servicio de movilidad urbana y ecologica, esta empresa es mediana y ya esta siendo reonocida e implementada fuera de Lima, Peru, se esta extendiendo en territorio estadounidense.

---

## ¿Qué datos genera?

- **Tipo:**  
  Datos geoespaciales (coordenadas GPS de inicio y fin), marcas de tiempo, ID de usuarios, estado mecanico de las bicicletas (bateria, frenos y posibilidad de adjuntas imagenes en caso de daños) y datos meteorológicos vinculados.

- **Volumen:**  
  Gigabytes de datos mensuales provenientes de miles de viajes diarios que registran la ubicación cada pocos segundos.

- **Frecuencia:**  
  Datos en tiempo real (*streaming*) transmitidos por los sensores IoT de cada bicicleta activa y por la aplicación móvil de los usuarios. [1]

---

## ¿Quién los usa?

- **Operadores de la empresa:**  
  Equipos de logística y mantenimiento mecánico.

- **Gobierno local:**  
  Municipalidades para gestionar la construccion de ciclovias.

- **Usuarios finales:**  
  Clientes que consumen datos procesados para ver la disponibilidad de bicicletas en tiempo real.

---

## ¿Qué problema resuelve?

### Logística de balanceo
Predice qué estaciones se quedarán vacías o llenas para mover las bicicletas de forma proactiva en furgonetas.

### Mantenimiento predictivo
Detecta fallas en los sensores antes de que la bicicleta quede en mal estado o como perdida total, reduciendo costos de reparación.

### Planificación urbana
Aporta datos precisos a la ciudad sobre cuáles son las rutas más transitadas para priorizar la construcción de infraestructura vial segura.

---

## ¿Por qué no bastaría Excel?

### Velocidad de flujo
Excel no puede procesar miles de coordenadas GPS que ingresan en tiempo real por segundo (*streaming*).

### Volumen límite
Una hoja de cálculo colapsaría rápidamente al superar el millón de filas, límite que este sistema excede en pocas semanas de operación.

### Variedad y complejidad
El análisis requiere combinar datos comunes de una base datos y datos mas pesados como json enteros, imagenes en binarios y todo lo que generan los equipos que conforman las bicicletas algo que una base de datos relacional simple o Excel no pueden optimizar eficientemente.
