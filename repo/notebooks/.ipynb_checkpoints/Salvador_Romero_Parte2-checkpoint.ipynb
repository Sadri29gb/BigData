# Ejercicio 16: DataFrame de estudiantes

## Objetivos
- Crear un DataFrame con información de estudiantes.
- Mostrar las primeras 5 filas.
- Mostrar información general del DataFrame.
- Calcular el promedio general de calificaciones.
- Mostrar estudiantes con promedio mayor o igual a 80.
- Ordenar estudiantes por promedio de mayor a menor.

```python

import pandas as pd
print("\n Ejercicio 16")
df_est = pd.DataFrame({
    'nombre': ['Ana', 'Luis', 'Carlos', 'Maria', 'Pedro', 'Sofia', 'Jorge', 'Elena', 'Diego', 'Valeria'],
    'carrera': ['Sistemas', 'Medicina', 'Derecho', 'Arquitectura', 'Sistemas', 'Administracion', 'Psicologia', 'Derecho', 'Medicina', 'Civil'],
    'edad': [20, 22, 19, 21, 23, 20, 24, 19, 22, 21],
    'promedio': [85, 92, 78, 88, 95, 70, 82, 90, 86, 79],
    'ciudad': ['Tgs', 'SPS', 'Yoro', 'CMY', 'SIG', 'La Paz', 'SNTRS', 'Copan', 'Progreso', 'LaEsperanza']
})


print("1. Primeras 5 filas:")
print(df_est.head())

print("\n2. Información general:")
print(df_est.info())

print("\n3. Promedio general:")
print(df_est['promedio'].mean())

print("\n4. Estudiantes con promedio mayor a 80:")
print(df_est[df_est['promedio'] >= 80])

print("\n5. Ordenados por promedio (mayor a menor):")
print(df_est.sort_values('promedio', ascending=False))
```

# Ejercicio 17: Análisis de ventas con pandas

## Objetivos
- Crear un DataFrame de ventas.
- Calcular una nueva columna llamada TotalVenta.
- Obtener el total vendido por vendedor.
- Identificar el producto más vendido.
- Identificar la categoría con mayor ingreso.
- Ordenar las ventas de mayor a menor.

```python

print("\n Ejercicio 17")
df_ven = pd.DataFrame({
    'fecha': pd.date_range('2024-01-01', periods=15),
    'vendedor': ['Ana', 'Luis', 'Ana', 'Carlos', 'Luis', 'Carlos', 'Ana', 'Luis', 'Carlos', 'Ana', 'Luis', 'Carlos', 'Ana', 'Luis', 'Carlos'],
    'producto': ['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Laptop', 'Mouse', 'Teclado', 'Monitor', 'Laptop', 'Mouse', 'Teclado', 'Monitor', 'Laptop', 'Mouse', 'Teclado'],
    'categoria': ['Tecnologia', 'Accesorios', 'Accesorios', 'Tecnologia', 'Tecnologia', 'Accesorios', 'Accesorios', 'Tecnologia', 'Tecnologia', 'Accesorios', 'Accesorios', 'Tecnologia', 'Tecnologia', 'Accesorios', 'Accesorios'],
    'cantidad': [2, 5, 10, 1, 3, 8, 12, 2, 4, 6, 9, 3, 1, 7, 11],
    'precio_unitario': [1000, 20, 50, 300, 1000, 20, 50, 300, 1000, 20, 50, 300, 1000, 20, 50]
})

df_ven['TotalVenta'] = df_ven['cantidad'] * df_ven['precio_unitario']
print("1. Total vendido por vendedor:")
print(df_ven.groupby('vendedor')['TotalVenta'].sum())

print("\n2. Producto más vendido (por cantidad):")
print(df_ven.groupby('producto')['cantidad'].sum().idxmax())

print("\n3. Categoría con mayor ingreso:")
print(df_ven.groupby('categoria')['TotalVenta'].sum().idxmax())

print("\n4. Ventas ordenadas de mayor a menor:")
print(df_ven.sort_values('TotalVenta', ascending=False))
```

# Ejercicio 18: Limpieza de datos con pandas

## Objetivos
- Crear un DataFrame con datos incompletos.
- Identificar valores nulos.
- Contar valores nulos por columna.
- Reemplazar ciudades vacías.
- Eliminar registros sin correo.
- Mostrar el DataFrame limpio.

```python
print("\n Ejercicio 18")
df_cli = pd.DataFrame({
    'nombre': ['Juan', 'Ana', None, 'Pedro', 'María', 'Luis', 'Sofia', 'Diego'],
    'edad': [25, 30, 22, None, 28, 35, 29, 31],
    'correo': ['juan@mail.com', None, 'pedro@mail.com', 'ana@mail.com', None, 'luis@mail.com', 'sofia@mail.com', 'diego@mail.com'],
    'ciudad': ['SPS', None, 'TGS', 'CMY', None, 'STRS', 'YORO', None],
    'telefono': [5551234, 5555678, None, 5559012, 5553456, 5557890, 5552345, 5556789]
})

print("1. Identificar null:")
print(df_cli.isna())

print("\n2. Conteo de null por columna:")
print(df_cli.isna().sum())

df_cli['ciudad'] = df_cli['ciudad'].fillna('No especificada')
df_cli = df_cli.dropna(subset=['correo'])

print("\n3. DataFrame limpio:")
print(df_cli)
```

# Ejercicio 19: Análisis de préstamos

## Objetivos
- Crear un DataFrame con solicitudes de préstamo.
- Calcular el monto promedio solicitado.
- Contar solicitudes por estado.
- Mostrar préstamos aprobados.
- Mostrar clientes con monto solicitado mayor a 50,000.
- Agrupar por estado y calcular el promedio del monto solicitado.

```python
print("\n Ejercicio 19")

df_prest = pd.DataFrame({
    'cliente': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'monto_solicitado': [20000, 60000, 15000, 80000, 45000, 55000, 30000, 70000],
    'plazo': [12, 24, 12, 36, 12, 24, 12, 36],
    'ingreso_mensual': [3000, 8000, 2500, 9000, 4000, 7500, 3500, 8500],
    'estado_solicitud': ['Aprobado', 'Rechazado', 'Aprobado', 'Aprobado', 'Pendiente', 'Aprobado', 'Rechazado', 'Aprobado']
})

print("1. Monto promedio solicitado:")
print(df_prest['monto_solicitado'].mean())

print("\n2. Solicitudes por estado:")
print(df_prest['estado_solicitud'].value_counts())

print("\n3. Solo aprobados:")
print(df_prest[df_prest['estado_solicitud'] == 'Aprobado'])

print("\n4. Monto mayor 50k:")
print(df_prest[df_prest['monto_solicitado'] > 50000])

print("\n5. Promedio de monto por estado:")
print(df_prest.groupby('estado_solicitud')['monto_solicitado'].mean())

```

# Ejercicio 20: Mini análisis de datos tipo empresa

## Objetivos
- Crear un DataFrame con información de empleados.
- Mostrar registros iniciales.
- Obtener estadísticas de salarios.
- Calcular salario promedio por departamento.
- Encontrar el empleado con mayor salario.
- Mostrar empleados con más de 5 años de experiencia.
- Ordenar empleados por salario.
- Crear una columna NivelExperiencia.

```python
df_emp = pd.DataFrame({
    'nombre': ['E1', 'E2', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8'],
    'departamento': ['TI', 'Ventas', 'TI', 'RRHH', 'Ventas', 'TI', 'RRHH', 'Ventas'],
    'cargo': ['Dev', 'Vendedor', 'QA', 'Analista', 'Gerente', 'Dev', 'Coordinador', 'Vendedor'],
    'salario': [5000, 3500, 4500, 4000, 6000, 5500, 4200, 3800],
    'años_experiencia': [1, 4, 6, 3, 8, 2, 5, 10],
    'ciudad': ['CMY', 'TGS', 'SPS', 'LaEspernza', 'SPS', 'SIG', 'Copan', 'SnraBrbara']
})

print("1. Primeros registros:") 
print(df_emp.head()) 
print("\n2. estadistica de salarios:") 
print(df_emp['salario'].describe()) 
print("\n3. Salario promedio por departamento:") 
print(df_emp.groupby('departamento')['salario'].mean())

print("\n4. Empleado con mayor salario:") 
print(df_emp.loc[df_emp['salario'].idxmax()]) 

print("\n5. Más de 5 años de experiencia:") 
print(df_emp[df_emp['años_experiencia'] > 5])

print("\n6. Ordenados por salario:") 
print(df_emp.sort_values('salario')) 

def nivel(anios):
    if anios <= 2:
        return 'Junior'
    elif anios <= 5:
        return 'Intermedio'
    else:
        return 'Senior' 
            
df_emp['NivelExperiencia'] = df_emp['años_experiencia'].apply(nivel)
print("\n7. DataFrame con nueva columna:") 
print(df_emp)
```
